```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
from sklearn.ensemble import RandomForestClassifier
```

```python
df = pd.read_csv("Iris.csv")
```

```python
df
```

```python
x = df.iloc[ : , : -1]

y = df.iloc[ : , -1 ]
```

```python
x
```

```python
y
```

```python
#Choosing the model 
model = RandomForestClassifier()
model.fit(x,y)
```

```python
#predictions of model 
predictions = model.predict(x)
```

```python
predictions
```

```python
# checking the preciction score
actual = y 
predicted = predictions
count = 0 
for i in range(len(df)):
    if actual[i] == predicted[i]:
        count +=1

print(f'Accuracy of the model : {(count*100)/len(df)}')
```

```python
test_data =[[11	,5.4	,3.7	,1.5,	0.2]]
model.predict(test_data)
```
