```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import StratifiedShuffleSplit
```

```python
# loading data 
data = pd.read_csv("Iris.csv")
```

```python
data
```

```python
split = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=42)

for train_index, test_index in split.split(data, data["Species"]):
    strat_train_set = data.loc[train_index]
    strat_test_set = data.loc[test_index]
```

```python
strat_train_set
```

```python
X_train = strat_train_set.iloc[: , : -1]
y_train = strat_train_set.iloc[ : , -1]
```

```python
X_train
```

```python
y_train
```

```python
model = RandomForestClassifier()
model.fit(X_train,y_train)
```

```python
X_test = strat_test_set.iloc[:,:-1]
y_test = strat_test_set.iloc[: , -1]
```

```python
#predict 
predict = model.predict(X_test)
print(predict)
```

```python
y_test
```

```python
actual = y_test.to_numpy()
predicted = predict

count = 0

for i in range(len(actual)):
    if actual[i] == predicted[i]:
        count += 1

print(f"Accuracy: {count * 100 / len(actual):.2f}%")
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, predict)

print(f"Accuracy: {accuracy:.2%}")
```
